# 2. Download Dataset & Train Smile Classification Model
Downloads your dataset from Roboflow and trains a YOLOv8 Nano Classification model for 50 epochs.

```python
from roboflow import Roboflow
from ultralytics import YOLO

# 1. Download your exact dataset (Folder format for classification)
rf = Roboflow(api_key="bavYV7hjjQK9oCPCzBTy")
project = rf.workspace("sheep-classification").project("smile-htr9e")
version = project.version(1)
dataset = version.download("folder")

# 2. Load the base YOLOv8 Nano Classification model
model = YOLO('yolov8n-cls.pt') 

# 3. Train the model (Uses Mac CPU/GPU automatically)
results = model.train(
    data=dataset.location, 
    epochs=50, 
    imgsz=64, 
    batch=32,
    project="trained_models",
    name="smile_model"
)
```

# 3. Export to ONNX
Exports the PyTorch `.pt` model to highly optimized ONNX format for lightning-fast inference on your Mac.

```python
from google.colab import files

# Export the best trained weights to ONNX format
onnx_path = model.export(format="onnx", imgsz=64)
print(f"\n[SUCCESS] Model successfully exported to: {onnx_path}")

# Download the file to your computer
files.download(onnx_path)
```
