# 2. Download Dataset & Train Model
Downloads your dataset from Roboflow and trains a YOLOv8 Nano Object Detection model for 50 epochs.

```python
from roboflow import Roboflow
from ultralytics import YOLO

# 1. Download your exact dataset
rf = Roboflow(api_key="bavYV7hjjQK9oCPCzBTy")
project = rf.workspace("abdul-mannan").project("thumbs-p8zym")
version = project.version(7)
dataset = version.download("yolov8")

# 2. Load the base YOLOv8 Nano Object Detection model
model = YOLO('yolov8n.pt') 

# 3. Train the model (Uses Mac CPU/GPU automatically)
results = model.train(
    data=f"{dataset.location}/data.yaml", 
    epochs=50, 
    imgsz=640, 
    batch=32,
    project="trained_models",
    name="thumbs_model"
)
```

# 3. Export to ONNX
Exports the PyTorch `.pt` model to highly optimized ONNX format for lightning-fast inference on your Mac.

```python
# Export the best trained weights to ONNX format
onnx_path = model.export(format="onnx", imgsz=640)
print(f"\n[SUCCESS] Model successfully exported to: {onnx_path}")
```
